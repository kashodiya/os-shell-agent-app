# Shell Agent Jupyter Notebook Example

This notebook demonstrates how to use the Shell Agent in a Jupyter notebook environment.

```python
import os
import sys

# Add the parent directory to the Python path
sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), '..')))
```

```python
# Import the shell agents
from src.shell_agent import ShellAgent
from src.advanced_shell_agent import AdvancedShellAgent
```

## Basic Shell Agent

First, let's try the basic shell agent to execute a simple command.

```python
# Get model ID from environment variable or use default
model_id = os.environ.get("STRANDS_MODEL_ID")

# Create the basic shell agent
basic_agent = ShellAgent(model_id=model_id)
```

```python
# Run a simple task
basic_result = basic_agent.run("list all files in the current directory")
print(basic_result)
```

## Advanced Shell Agent

Now, let's try the advanced shell agent to execute a more complex task.

```python
# Create the advanced shell agent
advanced_agent = AdvancedShellAgent(model_id=model_id, verbose=True)
```

```python
# Run a complex task
advanced_result = advanced_agent.run(
    "create a Python script that generates the first 10 prime numbers and run it"
)
print(advanced_result)
```

## Custom Task

Try your own custom task below:

```python
# Define your custom task
custom_task = "your custom task here"

# Choose which agent to use (basic or advanced)
use_advanced = True

# Run the task
if use_advanced:
    result = advanced_agent.run(custom_task)
else:
    result = basic_agent.run(custom_task)

print(result)
```
